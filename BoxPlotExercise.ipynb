```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
# boxplot
# median, quartile(25%, 50%, 75%, min max 1.5 * IQR, outlier
```

```python
data = np.array([5,7,9,15,20,22,25,30,32,35,37,40,50,55,60,100])

plt.figure(figsize=(6,5))
sns.boxplot(y=data)
plt.title("Box Plot")
plt.ylabel("Data Value")
plt.grid(False)
plt.show()
```

```python
data = np.array([5,7,9,15,20,22,25,30,32,35,37,40,50,55,60]) # outlier çıkarılırsa.

plt.figure(figsize=(6,5))
sns.boxplot(y=data)
plt.title("Box Plot")
plt.ylabel("Data Value")
plt.grid(False)
plt.show()
```

```python
df = sns.load_dataset("titanic")
```

```python
plt.figure(figsize=(15,13))

plt.subplot(1, 2, 1)
sns.boxplot(data=df, x="class", y="age")
plt.title("Age by Class")
plt.xlabel("Class")
plt.ylabel("Age")
plt.subplot(1, 2, 2)
sns.boxplot(data=df, x="class", y="fare")
plt.title("Fare by Class")
plt.xlabel("Class")
plt.ylabel("Fare")

plt.tight_layout()
plt.show()
```
