```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("athlete_events.csv")
```

```python
data.describe()
```

```python
data.head()
```

```python
data.columns
```

```python
data.info()
```

```python
plt.scatter("Height", "Weight", data=data)
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Athelete Weight vs Height")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.scatterplot(data=data, x="Height", y="Weight", hue="Sex")
plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Atheletes Weight vs Height")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.scatterplot(data=data, x="Height", y="Weight", hue="Medal")
plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Atheletes Weight vs Height")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.scatterplot(data=data, x="Height", y="Weight", hue="Sex", style="Medal")
plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Atheletes Weight vs Height")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.scatterplot(data=data, x="Height", y="Weight", hue="Sex", style="Medal", size="Age")
plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Atheletes Weight vs Height")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.lineplot(data=data, x="Height", y="Weight", hue="Sex")
plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Atheletes Weight vs Height")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.displot(data=data, x="Height", hue="Sex")
plt.ylabel("Frequency")
plt.title("Atheletes Height Distrubition")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.displot(data=data, x="Height", hue="Sex", kind="kde") # kde for kernal density
plt.ylabel("Frequency")
plt.title("Atheletes Height Distrubition")
plt.show()
```

```python
sns.barplot(data=data, x="Medal", y="Height", hue="Sex")
plt.title("Medals by Height")
plt.show()
```

```python
sns.catplot(data=data, x="Medal", y="Height", hue="Sex", col="Season")
plt.show()
```
