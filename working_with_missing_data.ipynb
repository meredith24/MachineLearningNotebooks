```python
import seaborn as sns
import pandas as pd
```

```python
df = sns.load_dataset("titanic")
```

```python
df.isnull().sum()
```

```python
# mean imputation

sns.histplot(data=df, x="age", kde=True)
```

```python
df["Age_mean"] = df["age"].fillna(df["age"].mean())
```

```python
df[["Age_mean", "age"]]
```

```python
df["age_median"] = df["age"].fillna(df["age"].median()) # fazla outlier olduğu için medyanla da ampute edebilirim.
```

```python
df[["age", "age_median", "Age_mean"]]
```

```python
df.dropna(subset=["embarked"], inplace=True)
```

```python
df.drop(columns=["deck"], inplace=True)
```

```python
df.isnull().sum()
```

```python
df.drop(columns=["age"], inplace=True)
```

```python
df.isnull().sum()
```

```python
df.drop(columns=["Age_mean"], inplace=True)
df.rename(columns={"age_median": "age"}, inplace=True)
```

```python
df.isnull().sum()
```

```python
sns.histplot(data=df, x="age", kde=True)
```
