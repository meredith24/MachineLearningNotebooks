```python
def salary_category(salary):
    if salary < 50000:
        return "low"
    elif 50000 <= salary < 80000:
        return "medium"
    else:
        return "high"
```

```python
df["Salary"].apply(salary_category)
```

```python
df["Salary_Category"] = df["Salary"].apply(salary_category)
```

```python
df
```

```python
def boost_performance_score(performance_score):
    performance_score += 1
    return performance_score
```

```python
experienced_df = df[df["Experience"] > 10]
```

```python
df["New_Performance_Score"] = experienced_df["Performance_Score"].apply(boost_performance_score)
```

```python
df
```

```python
df.drop("New_Performance_Score", axis=1, inplace=True)
```

```python
def adjust_performance_score(row):
    if row["Experience"] > 10:
        return row["Performance_Score"] + 1
    else:
        return row["Performance_Score"]
    
```

```python
df["Adjusted_Score"]= df.apply(adjust_performance_score, axis=1)
```

```python
df["Formatted_Name"] = df["Name"].apply(lambda x: x.replace("_", " "))
```
