```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
%matplotlib inline
```

```python
numpy_array1 = np.linspace(0, 20, 10)
numpy_array2 = numpy_array1 ** 3
```

```python
plt.plot(numpy_array1, numpy_array2, "r-*")
plt.show()
```

```python
plt.plot(numpy_array2, numpy_array1, "b-*")
plt.show()
```

```python
plt.subplot(1, 2, 1)
plt.plot(numpy_array1, numpy_array2, "r-*")
plt.subplot(1, 2, 2)
plt.plot(numpy_array2, numpy_array1, "g--")
plt.show()
         
```

```python
my_figure = plt.figure() # kendi grafiğimi oluşturuyorum.
figure_axes = my_figure.add_axes([0.1, 0.1, 0.3, 0.8]) # üstten ve alttan boşluk, genişlik ve yükseklik kontrolü
figure_axes.plot(numpy_array1, numpy_array2, "g")
figure_axes.set_xlabel("X Axis")
figure_axes.set_ylabel("Y Axis")
figure_axes.set_title("Graph Title")
plt.show()
```

```python
new_fig =  plt.figure(dpi=100) # yeni grafik nesnesi oluşturma ve çözünürlüğü ayarlama.
new_axes = new_fig.add_axes([0.1, 0.1, 0.6, 0.6])
new_axes.plot(numpy_array1, numpy_array1 ** 2, label="numpy array **2")
new_axes.plot(numpy_array1, numpy_array1 ** 3, label="numpy array **3")
new_axes.legend()
plt.show()
```

### Matplotlib Stilleri

```python
data1 = np.linspace(0, 10, 20)
data2 = data1 ** 2
```

```python
fig, axes = plt.subplots()
axes.plot(data1, data2, color="#5ED2DB") # başındaki hashtag işaretini unutmamalı.
axes.plot(data2, data1, color="#00F100")
plt.show()
```

```python
new_fig, new_axes = plt.subplots() # tuple unpacking.

new_axes.plot(data1, data1 + 2, color="blue", linewidth=1.0)
new_axes.plot(data1, data1 + 3, color="yellow", linewidth=3.0)
new_axes.plot(data1, data1 + 4, color="#A3D505", linestyle="-.")
new_axes.plot(data1, data1 + 8, color="black", marker="+", markersize=8, markerfacecolor="red")



plt.show()
```

```python
# scatter plots, histograms, box plots

plt.scatter(data1, data2)
plt.show()

```
